crie uma nova lista com os valores somente iguais ou maiores que 6

```python
# Minha resolução

lista = [8,5,6,3,9]
resultado = []

for n in range(0, len(lista)):
    if lista[n] >= 6 :
        resultado.append(lista[n])
print(resultado)
```

```python
# Resultado do gabarito

lista = [8,5,6,3,9]
resultado = []

for n in lista:
    if n >= 6:
        resultado.append(n)
print(resultado)
```

```python
# Resultado recomendado pelo gemini

lista = [8,5,6,3,9]
resultado = [n for n in lista if n >= 6]
print(resultado)
```
